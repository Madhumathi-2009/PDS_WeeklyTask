# Task 2 : Uivariate Analysis

```python
airline_count = df["airline"].value_counts().reset_index()

airline_count.columns = ["airline", "count"]

plt.figure(figsize=(8, 5))

sns.barplot(
    data=airline_count,
    x="airline",
    y="count"
)

plt.title("Number of Flights by Airline")
plt.xlabel("Airline")
plt.ylabel("Number of Flights")

plt.show()
```

```python
class_count = df["class"].value_counts().reset_index()

class_count.columns = ["class", "count"]

class_count

plt.figure(figsize=(3, 5))

sns.barplot(
    data=class_count,
    x="class",
    y="count"
)

plt.title("Number of Flights by Class")
plt.xlabel("Class")
plt.ylabel("Number of Flights")

plt.show()
```

```python
plt.figure(figsize=(4, 5))

sns.countplot(
    data=df,
    x="stops"
)

plt.title("Number of Flights by Stops")
plt.xlabel("Stops")
plt.ylabel("Number of Flights")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="duration",
    bins=30,
    kde=True
)

plt.title("Distribution of Flight Duration")
plt.xlabel("Flight Duration")
plt.ylabel("Number of Flights")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="price",
    bins=30,
    kde=True
)

plt.title("Distribution of Flight Prices")
plt.xlabel("Price")
plt.ylabel("Number of Flights")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(
    data=df,
    x="duration"
)

plt.title("Box Plot of Flight Duration")
plt.xlabel("Flight Duration")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(
    data=df,
    x="price"
)

plt.title("Box Plot of Flight Prices")
plt.xlabel("Price")

plt.show()
```

# Task 3 : Bivariate Analysis

```python
airline_price = df.groupby("airline", as_index=False)["price"].mean()

airline_price
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=airline_price,
    x="airline",
    y="price"
)

plt.title("Average Flight Price by Airline")
plt.xlabel("Airline")
plt.ylabel("Average Price")

plt.show()
```

```python
class_price = df.groupby("class", as_index=False)["price"].mean()

class_price
```

```python
plt.figure(figsize=(3, 5))

sns.barplot(
    data=class_price,
    x="class",
    y="price"
)

plt.title("Average Flight Price by Class")
plt.xlabel("Class")
plt.ylabel("Average Price")

plt.show()
```

```python
stops_price = df.groupby("stops", as_index=False)["price"].mean()

stops_price
```

```python
plt.figure(figsize=(4, 5))

sns.barplot(
    data=stops_price,
    x="stops",
    y="price"
)

plt.title("Average Flight Price by Stops")
plt.xlabel("Stops")
plt.ylabel("Average Price")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="duration",
    y="price",
)

plt.title("Flight Duration vs Price")
plt.xlabel("Flight Duration")
plt.ylabel("Price")

plt.show()
```

```python
days_price = df.groupby("days_left", as_index=False)["price"].mean()

days_price.head()
```

```python
plt.figure(figsize=(8, 5))

sns.lineplot(
    data=days_price,
    x="days_left",
    y="price",
    marker="o"
)

plt.title("Average Flight Price vs Days Left")
plt.xlabel("Days Left")
plt.ylabel("Average Price")

plt.show()
```

# Task 4 : Correlation Analysis & Feature Engineering

```python
corr = df[["duration", "days_left", "price"]].corr()

corr
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
df["price_per_hour"] = df["price"] / df["duration"]

df[["price", "duration", "price_per_hour"]].head()
```

```python
df["route"] = df["source_city"] + " to " + df["destination_city"]

df[["source_city", "destination_city", "route"]].head()
```

```python
df["stops"].unique()
```

```python
stops_mapping = {
    "zero": 0,
    "one": 1,
    "two_or_more": 2
}

df["stops_num"] = df["stops"].map(stops_mapping)

df[["stops", "stops_num"]].head()
```

```python
df[[
    "price_per_hour",
    "route",
    "stops_num"
]].head()
```
