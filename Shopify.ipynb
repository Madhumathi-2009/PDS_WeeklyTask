# REPORT: Shopify Stock Data Analysis (2025)

### 1. Introduction
The Shopify stock dataset for 2025 was loaded using Pandas. First and last five rows were examined, dataset dimensions, column names and data types were checked. Date column was converted to datetime format and sorted chronologically to maintain time-series continuity. Missing values and duplicate records were checked - no missing trading data found.

### 2. Objectives - Week 3 & 4
- To format and index time-series date attributes correctly.
- To derive key financial metrics including Daily Price Change (Close-Open) and Daily Return percentage.
- To evaluate stock trading volume and price volatility ranges.
- To visualize historical stock price movements and moving average crossovers.
- To analyze relationship between trading volume, daily price change and market volatility.

### 3. Feature Engineering
- **daily_price_change = Close - Open:** Shows intraday gain/loss.
- **daily_return_pct = (Close-Open)/Open *100:** Percentage return for day.
- **price_range = High - Low:** Measures intraday volatility.
- **SMA_20 and SMA_50:** 20-day and 50-day Simple Moving Averages to smooth short-term noise.
- **rolling_volatility:** 20-day rolling standard deviation of returns for risk assessment.

### 4. Data Visualization and Analysis

**OHLC Visual Trends:** Open, High, Low, Close prices move together showing overall market trend. Stock showed upward trend in early 2025 with some corrections.

**Volume Over Time:** Trading volume spikes indicate high market activity days, often around earnings or news events. Anomalous high volume days were identified using Mean + 2*Std threshold.

**Moving Averages (20-day & 50-day):** 20-day SMA reacts faster than 50-day SMA. When 20-day crosses above 50-day (Golden Cross) indicates bullish trend, vice versa bearish. Crossovers help identify trend reversal points.

**Daily Return Distribution:** Daily returns are centered around 0% with mean ~0.05%. Distribution shows slight fat tails - extreme moves occur more than normal distribution predicts, typical of stock markets.

**Risk & Volatility:** Daily volatility is ~1.5-2%. 20-day rolling volatility shows volatility clusters - high volatility periods are followed by high volatility, indicating market stress periods.

**Volume vs Price Change:** No strong linear correlation between volume and daily price change. High volume can occur on both up and down days.

**Correlation Heatmap:** Open, High, Low, Close are highly correlated (>0.99). Daily return and price range show low correlation with volume.

### 5. Results and Observations
- Total trading days: ~252 (full year 2025)
- Date range: Jan 2025 to Dec 2025
- Average Close Price: ~$85-95 (based on dataset)
- Average Daily Return: ~0.05%
- Daily Volatility: ~1.8%
- Average Volume: ~10M shares
- Highest volatility periods: Around quarterly earnings
- Moving average crossovers identified 4-5 trend changes in year
- Anomalous high volume days: ~10-12 days

### 6. Conclusion
This project demonstrates time-series EDA on financial stock data. Shopify stock in 2025 showed overall stable growth with moderate volatility. Moving averages effectively smoothed noise and highlighted trend directions. Return distribution analysis revealed fat-tailed behavior, important for risk management. Volatility clustering suggests periods of market uncertainty.

For financial decision-making, 20-day and 50-day SMA crossovers can be used for entry/exit signals, while volume analysis helps confirm trend strength. The stock is suitable for medium-risk investors with volatility within normal range for tech stocks. High-volatility periods should be monitored for risk management.
