# 9. To perform indexing, slicing and reshaping in NumPy

```python
import numpy as np                                # Importing NumPy library
a = np.array([5, 10, 15, 20, 25, 30])             # Creating a 1D NumPy array 
print("1D Array:", a,"\n")                        # Displaying the  1D array
print("Indexing:")                                # Accessing specific elements using indexing
print("Element at index 0:", a[0])
print("Element at index 3:", a[3])
print("Last element      :", a[-1],"\n")
print("Slicing:")                                 # Extract subsets using slicing
print("Elements from index 1 to 4:", a[1:5])
print("First 3 elements          :", a[:3])
print("Last 2 elements           :", a[-2:])
print("Every 2nd element         :", a[::2],"\n")
b = np.array([[1, 2, 3],                          # Creating a 2D NumPy array
              [4, 5, 6], 
              [7, 8, 9]])
print("2D Array:\n",b,"\n")                       # Displaying the 2D array
print("Accessing 2D elements:")                   # Accessing individual elements from 2D array
print("Element at row 0, col 1:", b[0, 1])
print("Element at row 2, col 2:", b[2, 2],"\n")
print("Slicing 2D array:")                        # Extracting rows and columns from 2D array
print("First row    :",b[0, :])
print("Second column:",b[:, 1])
print("Sub-matrix [0:2, 1:3]:\n",b[0:2, 1:3],"\n")
arr_reshape = np.array([1, 2, 3, 4, 5, 6])        # Creating another 1D array for reshaping
print("Original 1D array for reshaping:", arr_reshape,"\n")
matrix_2x3 = arr_reshape.reshape(2, 3)            # Reshaping into 2x3 matrix
print("Reshaped to 2x3 matrix:\n", matrix_2x3,"\n")
matrix_3x2 = arr_reshape.reshape(3, 2)            # Reshaping into 3x2 matrix
print("Reshaped to 3x2 matrix:\n", matrix_3x2,"\n")
flattened = matrix_2x3.flatten()                  # Converting reshaped array back to 1D using flatten()
print("Flattened back to 1D:", flattened,"\n")
print("Array Attributes for 2x3 matrix:")         # Displaying array attributes
print("Shape     :", matrix_2x3.shape)            
print("Dimensions:", matrix_2x3.ndim)
print("Size      :", matrix_2x3.size)
print("Data type :", matrix_2x3.dtype)
```

# 12. To Handle Missing data using pandas

```python
import pandas as pd                                # Importing libraries
import numpy as np                                 # Importing numpy for NaN

data = {                                           # Creating DataFrame with missing values
    "Name" : ["Sakshi", "Rahul", "Ankit", "Priya"],
    "Marks" : [95, np.nan, 88, np.nan],
    "City" : ["Chennai", "Delhi", np.nan, "Mumbai"]
}
df = pd.DataFrame(data)
print("Original Data Frame :")                     # Displaying original DataFrame
print(df)
print("\nChecking Missing values :")               # Checking missing values using isnull()
print(df.isnull())
print("\nCount of Missing Values :")               # Counting missing values in each column
print(df.isnull().sum())
print("\nAfter filling with default value 0:")     # Filling all missing with default value
print(df.fillna(0))
df["Marks"] = df["Marks"].fillna(df["Marks"].mean()) # Filling Marks column with mean value
print("\nAfter filling Marks with mean:")
print(df)
print("\nForward fill:")                           # Forward filling missing values
print(df.ffill())
print("\nBackward fill:")                          # Backward filling missing values
print(df.bfill())
print("\nAfter removing rows with missing values:")# Removing rows containing missing values
print(df.dropna())
df.replace("Chennai", "CHENNAI", inplace=True)     # Replacing specified value
print("\nAfter replacing Chennai to CHENNAI:")
print(df)
print("\nDataFrame Info:")                         # Displaying DataFrame info
df.info()
```

# 13. To group and aggregate data in pandas

```python
import pandas as pd                                                     # Importing pandas library
# creating dataframe
data = {
    "Department" : ["HR", "IT", "HR", "IT", "Sales", "Sales"],
    "Employee" : ["Amti", "Rahul", "Priya", "Ankit", "sakshi", "Riya"],
    "Salary" : [30000, 50000, 35000, 60000, 45000, 40000],
    "Experience" : [2, 5, 3, 6, 4, 2]
}
df = pd.DataFrame(data)
print("Original DataFrame :")                                           # Displaying original DataFrame
print(df)
grouped = df.groupby("Department")                                      # Grouping based on Department column
# Sum Aggregation
print("\nSum of Salary by Department :")                                # Total salary for each department
print(grouped["Salary"].sum())
# Average Aggregation
print("\nAverage Salary by Department :")                               # Average salary for each department
print(grouped["Salary"].mean())
# Maximum Aggregation
print("\nMaximum Salary by Department :")                               # Max salary in each department
print(grouped["Salary"].max())
# Minimum Aggregation
print("\nMinimum Salary by Department :")                               # Min salary in each department
print(grouped["Salary"].min())
print("\nNumber of Employees in each Department :")                     # Counting employees in each department
print(grouped["Employee"].count())
print("\nMultiple Aggregations on Salary :")                            # Multiple aggregations sum, mean, max, min
print(grouped["Salary"].agg(['sum','mean','max','min']))
print("\nGrouped by Department and Experience :")                       # Grouping using Department and Experience
print(df.groupby(['Department','Experience'])['Salary'].sum())
print("\nIterating through each group:")                                # Iterating through each group
for name, group in df.groupby("Department"):
    print(f"\nDepartment: {name}")
    print(group)
print("\nDataFrame sorted by Salary :")                                 # Sorting DataFrame by Salary column
print(df.sort_values('Salary'))
```

# 14. To merge and join DataFrames using pandas

```python
import pandas as pd                                      # Importing pandas library
df1 = pd.DataFrame({                                     # Creating first DataFrame
    "ID" : [101, 102, 103, 104],
    "Name" : ["Sakshi", "Rahul", "Ankita", "Priya"]
})
print("DataFrame 1:")                                    # Displaying first DataFrame
print(df1)

df2 = pd.DataFrame({                                     # Creating second DataFrame
    "ID" : [101, 102, 103, 105],
    "Marks" : [95, 88, 90, 85]
})
print("\nDataFrame 2:")                                  # Displaying second DataFrame
print(df2)
print("\nInner Merge:")                                  # Inner merge on ID column
print(pd.merge(df1, df2, on="ID", how="inner"))
print("\nLeft Merge:")                                   # Left merge on ID column
print(pd.merge(df1, df2, on="ID", how="left"))
print("\nRight Merge:")                                  # Right merge on ID column
print(pd.merge(df1, df2, on="ID", how="right"))
print("\nOuter Merge:")                                  # Outer merge on ID column
print(pd.merge(df1, df2, on="ID", how="outer"))
df3 = pd.DataFrame({                                     # Creating DataFrames with index for join
    "Age" : [20, 21, 22]
}, index=[101, 102, 103])

df4 = pd.DataFrame({
    "City" : ["Chennai", "Delhi", "Mumbai"]
}, index=[101, 102, 103])
print("\nDataFrame 3:")                                  # Displaying new DataFrames for join
print(df3)
print("\nDataFrame 4:")
print(df4)
print("\nDefault Join:")                                 # Default join method
print(df3.join(df4))
print("\nLeft Join:")                                    # Left join operation
print(df3.join(df4, how="left"))
print("\nRight Join:")                                   # Right join operation
print(df3.join(df4, how="right"))
print("\nOuter Join:")                                   # Outer join operation
print(df3.join(df4, how="outer"))
```

# 15. To Create simple plots using Matplotlib (line, bar & histogram)

# 18. To perform basic data cleaning and preprocessing 

```python
import pandas as pd # Importing pandas library
import numpy as np # Importing numpy library

# Creating sample DataFrame 
data = { 
    "Name": ["Sakshi", "Rahul", "Priya", "Ankit", "Rahul", "Riya", None],
    "Age": [20, 21, np.nan, 22, 21, 20, 23],
    "Marks": [95, np.nan, 90, 85, 88, 92, 80],
    "Department": ["IT", "HR", None, "Sales", "HR", "IT", "HR"]
}

df = pd.DataFrame(data)                              # Creating DataFrame
print("Original Dataset:")                           # Displaying original dataset
print(df)

print("\nMissing Values in Each Column:")            # Checking missing values
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())       # Filling missing numeric values with mean
df["Marks"] = df["Marks"].fillna(df["Marks"].mean()) # Filling missing numeric values with mean

df["Name"] = df["Name"].fillna("Unknown")            # Filling missing categorical with default
df["Department"] = df["Department"].fillna("Unknown")# Filling missing categorical with default

print("\nAfter Filling Missing Values:")             # Displaying after filling missing values
print(df)

df.drop_duplicates(inplace=True)                     # Removing duplicate records
print("\nAfter Removing Duplicates:")                # Displaying after removing duplicates
print(df)

df["Age"] = df["Age"].astype(int)                    # Converting Age column to integer 
print("\nAfter Converting Age to int:")              # Displaying data types
print(df.dtypes)

df.rename(columns={"Marks": "Score"}, inplace=True) # Renaming Marks column to Score
print("\nAfter Renaming Marks to Score:")           # Displaying after renaming
print(df)

df_sorted = df.sort_values("Score", ascending=False)# Sorting in descending order by Score
print("\nDataset Sorted by Score (Descending):")    # Displaying sorted dataset
print(df_sorted)

print("\nRecords Where Score > 85:")                # Filtering records where Score > 85
print(df[df["Score"] > 85])

print("\nFinal Dataset Information:")               # Displaying final dataset info using info()
df.info()
```

# 19. To compute basic Statistics (mean, median, mode, std) from a dataset

```python
import pandas as pd                              # Importing pandas library
# Creating sample dataset containing students marks
data = { 
    "Name": ["Sakshi", "Rahul", "Priya", "Ankit", "Riya", "Aman", "Sneha"],
    "Marks": [95, 88, 90, 85, 88, 92, 80]
}

df = pd.DataFrame(data)                          # Converting dataset into DataFrame
print("Dataset:")                                # Displaying the dataset
print(df)

mean_marks = df["Marks"].mean()                  # Calculating mean of marks
print(f"\nMean of Marks: {mean_marks}")

median_marks = df["Marks"].median()              # Calculating median of marks
print(f"Median of Marks: {median_marks}")

mode_marks = df["Marks"].mode()                  # Calculating mode of marks
print(f"Mode of Marks: {mode_marks.values}")

std_marks = df["Marks"].std()                    # Calculating standard deviation of marks
print(f"Standard Deviation of Marks: {std_marks}")

min_marks = df["Marks"].min()                    # Finding minimum marks
max_marks = df["Marks"].max()                    # Finding maximum marks
print(f"Minimum Marks: {min_marks}")
print(f"Maximum Marks: {max_marks}")

variance_marks = df["Marks"].var()               # Calculating variance of marks
print(f"Variance of Marks: {variance_marks}")

print("\nStatistical Summary using describe():") # Displaying statistical summary
print(df["Marks"].describe())
```

# 20. To create data visualizations using Seaborn (Heatmap, Boxplot & Pairplot)

```python
import pandas as pd            # Importing pandas library
import seaborn as sns          # Importing seaborn library
import matplotlib.pyplot as plt # Importing matplotlib library

# ----creating dataset------ 
data={                        # Creating dataset for visualization
    "Maths":[85,90,78,92,88],
    "Science":[80,85,75,98,67],
    "Social":[78,88,72,90,84]
}

# ----creating data frame------
df=pd.DataFrame(data)         # Creating DataFrame
print("==========DATASET============")             # Displaying dataset
print(df)

#-------HEATMAP-----------
plt.figure(figsize=(5,4))     # Creating figure for heatmap
sns.heatmap(df, annot=True, cmap='coolwarm') # Creating heatmap with annotations
plt.title("HEATMAP")          # Adding title
plt.show()                    # Displaying heatmap

#----------BOXPLOT-------------
plt.figure(figsize=(5,4))     # Creating figure for boxplot
sns.boxplot(data=df)          # Creating boxplot
plt.title("BOXPLOT")          # Adding title
plt.show()                    # Displaying boxplot

#---------PAIRPLOT-------------
sns.pairplot(df)              # Creating pairplot
plt.show()                    # Displaying pairplot
```
