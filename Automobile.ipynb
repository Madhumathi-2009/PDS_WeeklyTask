#### Converting technical columns to numeric

```python
cols_to_numeric = ['normalized-losses','bore','stroke','horsepower','peak-rpm','price']
for col in cols_to_numeric:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df.dtypes
```

#### Handling missing values - impute with median by body-style

```python
df['normalized-losses'] = df['normalized-losses'].fillna(df['normalized-losses'].median())
df['bore'] = df['bore'].fillna(df['bore'].median())
df['stroke'] = df['stroke'].fillna(df['stroke'].median())
df['horsepower'] = df['horsepower'].fillna(df.groupby('body-style')['horsepower'].transform('median'))
df['peak-rpm'] = df['peak-rpm'].fillna(df['peak-rpm'].median())
df['price'] = df['price'].fillna(df.groupby('make')['price'].transform('median'))

df.isnull().sum()
```

```python
df.columns
```

#### Feature Engineering and Further Analysis

##### 1. Power-to-weight ratio

```python
df['power_to_weight'] = df['horsepower'] / df['curb-weight']
```

##### 2. Combined MPG - Average fuel consumption

```python
df['combined_mpg'] = (df['city-mpg'] + df['highway-mpg']) / 2
```

# REPORT: Exploratory Data Analysis on Automobile Dataset

### 1. Introduction
The dataset was loaded using Pandas. The first and last five rows were examined, and the dataset dimensions, column names, and data types were checked. Non-standard missing values '?' were replaced with NaN and converted to numeric types. Missing values were imputed using median and group-wise median to retain data integrity.

### 2. Objectives
- To detect and resolve non-standard missing values in technical specifications.
- To convert string-encoded numeric specs (horsepower, price, bore, stroke, peak-rpm) to proper numeric types.
- To engineer performance metrics like power-to-weight ratio and combined MPG.
- To analyze trade-offs between performance, efficiency and price for market positioning.

### 3. Data Visualization and Analysis

**Car Price Distribution:**
The price distribution is right-skewed. Most cars are priced between 10000-15000, with few premium cars above 30000. This indicates a budget-dominated market with a small luxury segment.

**Horsepower vs Price:**
There is a strong positive correlation. Higher horsepower leads to higher price. Performance is a key pricing driver.

**Engine Size vs City MPG Trade-off:**
There is a strong negative correlation. Larger engines have lower mileage, showing a clear performance vs efficiency trade-off. Buyers must choose between power and fuel economy.

**Price Based on Body Style:**
Convertible and hardtop have higher average price than hatchback and sedan. Body style significantly impacts pricing due to design and exclusivity.

**Fuel Type vs Price:**
Diesel and gas variants show price differences. Diesel cars have different horsepower and efficiency characteristics.

**Brand Positioning - Budget vs Premium:**
Brands like Jaguar, Mercedes-Benz, BMW, Porsche are in the premium segment (avg price > 25000), while Toyota, Honda, Nissan, Subaru are in the budget/mid-range segment. This segmentation helps identify market positioning.

**Correlation Heatmap:**
Engine-size (0.87), horsepower (0.81), curb-weight (0.77) and width strongly correlate with price. City-MPG (-0.68) and highway-MPG (-0.70) negatively correlate with price.

### 4. Feature Engineering
- **power_to_weight = horsepower / curb-weight:** Measures acceleration potential.
- **combined_mpg = (city-mpg + highway-mpg) / 2:** Measures average fuel efficiency.
- **price_segment:** Categorized into Budget (<10000), Mid-Range (10000-20000), Premium (>20000).

This project demonstrates how Exploratory Data Analysis can be applied to automobile data using Python libraries such as Pandas, Matplotlib, and Seaborn.

Visualizations revealed that automobile price is primarily driven by technical performance specifications like engine-size and horsepower, rather than just brand. The trade-off analysis between engine size and MPG highlights the engineering compromise between performance and fuel efficiency.
