# Task 2: Univariate Analysis

## Objective

Univariate analysis examines one variable at a time to understand its distribution, frequency, range, and possible outliers.

The following variables are analyzed:

- Airline
- Travel class
- Number of stops
- Flight duration
- Ticket price

Histograms, count plots, and box plots are used to visualize the data.

```python
# ============================================================
# TASK 2: UNIVARIATE ANALYSIS
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

# Set the graph style
sns.set_theme(style="whitegrid")

print("TASK 2: UNIVARIATE ANALYSIS")


# ------------------------------------------------------------
# 1. Histogram of Flight Duration
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="duration",
    bins=30,
    kde=True
)

plt.title("Distribution of Flight Duration")
plt.xlabel("Flight Duration (Hours)")
plt.ylabel("Number of Flights")

plt.show()


# ------------------------------------------------------------
# 2. Histogram of Ticket Price
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="price",
    bins=30,
    kde=True
)

plt.title("Distribution of Flight Ticket Prices")
plt.xlabel("Ticket Price")
plt.ylabel("Number of Flights")

plt.show()


# ------------------------------------------------------------
# 3. Count Plot of Airlines
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x="airline",
    order=df["airline"].value_counts().index
)

plt.title("Number of Flights by Airline")
plt.xlabel("Airline")
plt.ylabel("Number of Records")

plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# 4. Count Plot of Travel Class
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

sns.countplot(
    data=df,
    x="class",
    order=df["class"].value_counts().index
)

plt.title("Number of Flights by Travel Class")
plt.xlabel("Travel Class")
plt.ylabel("Number of Records")

plt.show()


# ------------------------------------------------------------
# 5. Count Plot of Number of Stops
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

sns.countplot(
    data=df,
    x="stops",
    order=df["stops"].value_counts().index
)

plt.title("Number of Flights by Number of Stops")
plt.xlabel("Number of Stops")
plt.ylabel("Number of Records")

plt.show()


# ------------------------------------------------------------
# 6. Box Plot of Flight Duration
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="duration"
)

plt.title("Box Plot of Flight Duration")
plt.xlabel("Flight Duration (Hours)")

plt.show()


# ------------------------------------------------------------
# 7. Box Plot of Ticket Price
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="price"
)

plt.title("Box Plot of Flight Ticket Prices")
plt.xlabel("Ticket Price")

plt.show()


# ------------------------------------------------------------
# 8. Summary Observations
# ------------------------------------------------------------

print("\nAIRLINE FREQUENCY:")
print(df["airline"].value_counts())

print("\nTRAVEL CLASS FREQUENCY:")
print(df["class"].value_counts())

print("\nNUMBER OF STOPS FREQUENCY:")
print(df["stops"].value_counts())

print("\nFLIGHT DURATION SUMMARY:")
print(df["duration"].describe())

print("\nTICKET PRICE SUMMARY:")
print(df["price"].describe())

print("\nTASK 2 COMPLETED SUCCESSFULLY!")
```

# Task 3: Bivariate Analysis

## Objective

Bivariate analysis examines the relationship between two variables.

In this task, ticket price is compared with:

- Airline
- Travel class
- Number of stops
- Flight duration
- Number of days left before departure

The purpose is to identify the factors that are associated with changes in airline ticket prices.

```python
# ============================================================
# TASK 3: BIVARIATE ANALYSIS
# ============================================================

print("TASK 3: BIVARIATE ANALYSIS")


# ------------------------------------------------------------
# 1. Ticket Price versus Airline
# ------------------------------------------------------------

plt.figure(figsize=(13, 6))

sns.boxplot(
    data=df,
    x="airline",
    y="price"
)

plt.title("Ticket Price by Airline")
plt.xlabel("Airline")
plt.ylabel("Ticket Price")

plt.xticks(rotation=45)

plt.show()


# ------------------------------------------------------------
# 2. Ticket Price versus Travel Class
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

sns.boxplot(
    data=df,
    x="class",
    y="price"
)

plt.title("Ticket Price by Travel Class")
plt.xlabel("Travel Class")
plt.ylabel("Ticket Price")

plt.show()


# ------------------------------------------------------------
# 3. Ticket Price versus Number of Stops
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="stops",
    y="price",
    order=df["stops"].value_counts().index
)

plt.title("Ticket Price by Number of Stops")
plt.xlabel("Number of Stops")
plt.ylabel("Ticket Price")

plt.show()


# ------------------------------------------------------------
# 4. Ticket Price versus Flight Duration
# ------------------------------------------------------------

# A sample is used so the graph loads faster

sample_size = min(10000, len(df))

sample_df = df.sample(
    n=sample_size,
    random_state=42
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=sample_df,
    x="duration",
    y="price",
    alpha=0.3
)

plt.title("Flight Duration versus Ticket Price")
plt.xlabel("Flight Duration (Hours)")
plt.ylabel("Ticket Price")

plt.show()


# ------------------------------------------------------------
# 5. Ticket Price versus Days Left
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.regplot(
    data=sample_df,
    x="days_left",
    y="price",
    scatter_kws={"alpha": 0.15},
    line_kws={"color": "red"}
)

plt.title("Days Left versus Ticket Price")
plt.xlabel("Days Left Before Departure")
plt.ylabel("Ticket Price")

plt.show()


# ------------------------------------------------------------
# 6. Average Price by Airline
# ------------------------------------------------------------

print("\nAVERAGE TICKET PRICE BY AIRLINE:")

average_price_airline = (
    df.groupby("airline")["price"]
    .mean()
    .sort_values(ascending=False)
)

print(
    average_price_airline.round(2)
)


# ------------------------------------------------------------
# 7. Average Price by Travel Class
# ------------------------------------------------------------

print("\nAVERAGE TICKET PRICE BY TRAVEL CLASS:")

average_price_class = (
    df.groupby("class")["price"]
    .mean()
    .sort_values(ascending=False)
)

print(
    average_price_class.round(2)
)


# ------------------------------------------------------------
# 8. Average Price by Number of Stops
# ------------------------------------------------------------

print("\nAVERAGE TICKET PRICE BY NUMBER OF STOPS:")

average_price_stops = (
    df.groupby("stops")["price"]
    .mean()
    .sort_values(ascending=False)
)

print(
    average_price_stops.round(2)
)


# ------------------------------------------------------------
# 9. Five Data-Based Insights
# ------------------------------------------------------------

highest_price_airline = average_price_airline.idxmax()
lowest_price_airline = average_price_airline.idxmin()

highest_price_class = average_price_class.idxmax()
lowest_price_class = average_price_class.idxmin()

highest_price_stops = average_price_stops.idxmax()
lowest_price_stops = average_price_stops.idxmin()

print("\nFIVE INSIGHTS:")

print(
    "1. The airline with the highest average ticket price is:",
    highest_price_airline
)

print(
    "2. The airline with the lowest average ticket price is:",
    lowest_price_airline
)

print(
    "3. The travel class with the highest average price is:",
    highest_price_class
)

print(
    "4. The travel class with the lowest average price is:",
    lowest_price_class
)

print(
    "5. The stop category with the highest average price is:",
    highest_price_stops
)

print(
    "\nAdditional observation:"
)

print(
    "The duration and days-left graphs show how flight duration "
    "and booking time are related to ticket price."
)

print("\nTASK 3 COMPLETED SUCCESSFULLY!")
```

# Task 4: Correlation and Feature Engineering

## Objective

The purpose of this task is to identify the features that are most related to ticket price and create new useful variables.

The following operations are performed:

- Encode categorical columns into numerical values
- Calculate correlations
- Create a correlation heatmap
- Identify the features most strongly related to ticket price
- Create new features from the existing data

```python
# ============================================================
# TASK 4: CORRELATION AND FEATURE ENGINEERING
# ============================================================

import numpy as np

print("TASK 4: CORRELATION AND FEATURE ENGINEERING")


# ------------------------------------------------------------
# 1. Create a Copy of the Dataset
# ------------------------------------------------------------

encoded_df = df.copy()


# ------------------------------------------------------------
# 2. Remove the Index Column if It Exists
# ------------------------------------------------------------

if "index" in encoded_df.columns:

    encoded_df = encoded_df.drop(
        columns=["index"]
    )


# ------------------------------------------------------------
# 3. Identify Categorical Columns
# ------------------------------------------------------------

categorical_columns = (
    encoded_df
    .select_dtypes(include=["object"])
    .columns
)

print("\nCATEGORICAL COLUMNS:")

print(
    categorical_columns.tolist()
)


# ------------------------------------------------------------
# 4. Encode Categorical Columns
# ------------------------------------------------------------

encoded_df = pd.get_dummies(
    encoded_df,
    columns=categorical_columns,
    drop_first=True
)


# ------------------------------------------------------------
# 5. Convert Boolean Columns to Integers
# ------------------------------------------------------------

boolean_columns = (
    encoded_df
    .select_dtypes(include=["bool"])
    .columns
)

encoded_df[boolean_columns] = (
    encoded_df[boolean_columns]
    .astype(int)
)


print("\nENCODED DATASET:")

display(
    encoded_df.head()
)


# ------------------------------------------------------------
# 6. Calculate Correlation with Price
# ------------------------------------------------------------

price_correlation = (
    encoded_df
    .corr()["price"]
    .sort_values(
        key=lambda x: abs(x),
        ascending=False
    )
)


print("\nFEATURES MOST RELATED TO PRICE:")

print(
    price_correlation.head(15)
)


# ------------------------------------------------------------
# 7. Correlation Heatmap
# ------------------------------------------------------------

top_features = (
    price_correlation
    .head(15)
    .index
)

plt.figure(figsize=(14, 10))

sns.heatmap(
    encoded_df[top_features].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title(
    "Correlation Heatmap: Top Features Related to Price"
)

plt.show()


# ------------------------------------------------------------
# FEATURE ENGINEERING
# ------------------------------------------------------------


# ------------------------------------------------------------
# New Feature 1: Duration Category
# ------------------------------------------------------------

df["duration_category"] = pd.cut(
    df["duration"],
    bins=[0, 3, 6, 12, np.inf],
    labels=[
        "Short",
        "Medium",
        "Long",
        "Very Long"
    ],
    include_lowest=True
)


# ------------------------------------------------------------
# New Feature 2: Booking Category
# ------------------------------------------------------------

df["booking_category"] = pd.cut(
    df["days_left"],
    bins=[0, 7, 30, np.inf],
    labels=[
        "Last Minute",
        "Early Booking",
        "Advance Booking"
    ],
    include_lowest=True
)


# ------------------------------------------------------------
# New Feature 3: Price Per Hour
# ------------------------------------------------------------

df["price_per_hour"] = (
    df["price"]
    / df["duration"]
)


# ------------------------------------------------------------
# New Feature 4: Direct Flight Indicator
# ------------------------------------------------------------

df["is_direct"] = (
    df["stops"]
    .astype(str)
    .str.lower()
    .eq("zero")
    .astype(int)
)


# ------------------------------------------------------------
# Display New Features
# ------------------------------------------------------------

print("\nNEW FEATURES CREATED:")

display(
    df[
        [
            "duration",
            "duration_category",
            "days_left",
            "booking_category",
            "price",
            "price_per_hour",
            "is_direct"
        ]
    ].head(10)
)


print("\nTASK 4 COMPLETED SUCCESSFULLY!")
```

# Task 5: Business Insights and Recommendations

## Objective

The purpose of this task is to convert the results of the data analysis into useful business findings.

The analysis provides:

- Five important findings from the dataset
- Insights into airline ticket pricing
- Three recommendations for airline companies

```python
# ============================================================
# TASK 5: BUSINESS INSIGHTS AND RECOMMENDATIONS
# ============================================================

print("TASK 5: BUSINESS INSIGHTS")


# ------------------------------------------------------------
# Calculate Important Results
# ------------------------------------------------------------

average_price_airline = (
    df.groupby("airline")["price"]
    .mean()
)

average_price_class = (
    df.groupby("class")["price"]
    .mean()
)

average_price_stops = (
    df.groupby("stops")["price"]
    .mean()
)


highest_airline = (
    average_price_airline.idxmax()
)

lowest_airline = (
    average_price_airline.idxmin()
)

highest_class = (
    average_price_class.idxmax()
)

lowest_class = (
    average_price_class.idxmin()
)

highest_stop_category = (
    average_price_stops.idxmax()
)

lowest_stop_category = (
    average_price_stops.idxmin()
)


# ------------------------------------------------------------
# Top Five Findings
# ------------------------------------------------------------

print("\nTOP FIVE FINDINGS")

print(
    "\n1. Travel Class:"
)

print(
    "The travel class with the highest average ticket price is",
    highest_class,
    "with an average price of",
    round(
        average_price_class[highest_class],
        2
    )
)


print(
    "\n2. Lower-Priced Travel Class:"
)

print(
    "The travel class with the lowest average ticket price is",
    lowest_class,
    "with an average price of",
    round(
        average_price_class[lowest_class],
        2
    )
)


print(
    "\n3. Airline Pricing:"
)

print(
    "The airline with the highest average ticket price is",
    highest_airline,
    "with an average price of",
    round(
        average_price_airline[highest_airline],
        2
    )
)


print(
    "\n4. Lower-Priced Airline:"
)

print(
    "The airline with the lowest average ticket price is",
    lowest_airline,
    "with an average price of",
    round(
        average_price_airline[lowest_airline],
        2
    )
)


print(
    "\n5. Number of Stops:"
)

print(
    "The stop category with the highest average ticket price is",
    highest_stop_category,
    "with an average price of",
    round(
        average_price_stops[highest_stop_category],
        2
    )
)


# ------------------------------------------------------------
# Business Recommendations
# ------------------------------------------------------------

print("\n" + "=" * 60)

print("THREE RECOMMENDATIONS FOR AIRLINES")

print("=" * 60)


print(
    "\n1. USE DYNAMIC PRICING"
)

print(
    "Airlines should adjust ticket prices according to "
    "booking time, travel class, demand, route, and "
    "flight characteristics."
)


print(
    "\n2. ENCOURAGE EARLY BOOKINGS"
)

print(
    "Airlines can provide early-booking discounts, "
    "special offers, or loyalty points to encourage "
    "customers to book in advance."
)


print(
    "\n3. CREATE CUSTOMER-SPECIFIC OFFERS"
)

print(
    "Airlines can provide different offers for Economy "
    "and Business customers and create suitable pricing "
    "for direct and connecting flights."
)


print("\n" + "=" * 60)

print(
    "AIRLINES FLIGHTS DATA ANALYSIS COMPLETED SUCCESSFULLY!"
)

print("=" * 60)
```
