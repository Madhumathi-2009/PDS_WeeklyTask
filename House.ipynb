#### Converting the Date Column

```python
df['date'] = pd.to_datetime(df['date'])

df['date'].head()
```

#### Extracting the Sale Year and Month

```python
df['sale_year'] = df['date'].dt.year
df['sale_month'] = df['date'].dt.month

df[['date', 'sale_year', 'sale_month']].head()
```

#### Checking House Prices

```python
df['price'].describe()
```

#### Distribution of House Prices

```python
plt.figure(figsize=(8,5))
sns.histplot(df['price'], bins=50, kde=True)

plt.title("Distribution of House Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

#### Distribution of Living Area

```python
plt.figure(figsize=(8,5))
sns.histplot(df['sqft_living'], bins=50, kde=True)

plt.title("Distribution of Living Area")
plt.xlabel("Living Area (Square Feet)")
plt.ylabel("Frequency")
plt.show()
```

#### House Price vs Living Area

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='sqft_living', y='price', data=df)

plt.title("House Price vs Living Area")
plt.xlabel("Living Area (Square Feet)")
plt.ylabel("House Price")
plt.show()
```

#### House Price Based on Grade

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='grade', y='price', data=df)

plt.title("House Price Based on Grade")
plt.xlabel("House Grade")
plt.ylabel("House Price")
plt.show()
```

#### Waterfront and House Price

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='waterfront', y='price', data=df)

plt.title("Waterfront vs House Price")
plt.xlabel("Waterfront (0 = No, 1 = Yes)")
plt.ylabel("House Price")
plt.show()
```

#### Checking the Relationship Between Numerical Columns

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```

#### Feature Engineering and Further Analysis

```python
df['house_age'] = df['sale_year'] - df['yr_built']

df[['yr_built', 'sale_year', 'house_age']].head()
```

#### Checking Renovated Houses

```python
df['renovated'] = np.where(df['yr_renovated'] > 0, 1, 0)

df[['yr_renovated', 'renovated']].head()
```

#### Calculating Price per Square Foot

```python
df['price_per_sqft'] = df['price'] / df['sqft_living']

df[['price', 'sqft_living', 'price_per_sqft']].head()
```

#### House Age vs House Price

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='house_age', y='price', data=df)

plt.title("House Age vs House Price")
plt.xlabel("House Age")
plt.ylabel("House Price")
plt.show()
```

#### Living Area vs House Price with Regression Line

```python
plt.figure(figsize=(8,5))
sns.regplot(x='sqft_living', y='price', data=df, scatter_kws={'alpha':0.3})

plt.title("Living Area vs House Price")
plt.xlabel("Living Area (Square Feet)")
plt.ylabel("House Price")
plt.show()
```

#### Checking Outliers in House Prices

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['price'])

plt.title("Outliers in House Prices")
plt.xlabel("House Price")
plt.show()
```

#### Checking Outliers in Living Area

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['sqft_living'])

plt.title("Outliers in Living Area")
plt.xlabel("Living Area (Square Feet)")
plt.show()
```

#### House Prices by Sale Year

```python
plt.figure(figsize=(8,5))
sns.lineplot(x='sale_year', y='price', data=df)

plt.title("House Prices by Sale Year")
plt.xlabel("Sale Year")
plt.ylabel("House Price")
plt.show()
```

#### House Locations

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x='long', y='lat', hue='price', data=df, palette='coolwarm', legend=False)

plt.title("Geographical Distribution of House Prices")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()
```

Exploratory Data Analysis (EDA) is used to understand a dataset through statistical summaries and visualizations. In this project, the House Sales Dataset from King County is analyzed to understand the factors that influence house prices.

The dataset was loaded using Pandas. The first and last five rows were examined, and the dataset dimensions, column names, and data types were checked. Missing values and duplicate records were also investigated.

New features were created to support the analysis:

- **House Age:** Calculated by subtracting the construction year from the sale year.

- **Price per Square Foot:** Calculated by dividing the house price by its living area.
